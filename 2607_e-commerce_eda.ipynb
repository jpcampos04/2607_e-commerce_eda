# Data cleaning

```python
# ========================
# Cleaning headers
# ========================

print(df_list_orders.columns.to_list())
print()
print(df_order_details.columns.to_list())
print()
print(df_sales_target.columns.to_list())
```

```python
df_list_orders.columns = df_list_orders.columns.str.strip().str.lower().str.replace(' ', '_')
df_list_orders = df_list_orders.rename(columns={'customername': 'customer_name'})

df_order_details.columns = df_order_details.columns.str.strip().str.lower().str.replace(' ', '_').str.replace('-', '_')

df_sales_target.columns = df_sales_target.columns.str.strip().str.lower().str.replace(' ', '_')

print(df_list_orders.columns.to_list())
print()
print(df_order_details.columns.to_list())
print()
print(df_sales_target.columns.to_list())
```

```python
# =====================================
# Looking for duplicates
# =====================================

print(df_list_orders.duplicated().sum(), 'Duplicates in dataset')
print()
print(df_order_details.duplicated().sum(), 'Duplicates in dataset')
print()
print(df_sales_target.duplicated().sum(), 'Duplicates in dataset')
```

```python
# =====================================
# Categorial typos
# =====================================

# Verify that there are no duplicates in the categorical columns
print(df_list_orders['order_id'].unique())
print()
print(df_list_orders['customer_name'].unique())
print()
print(df_list_orders['state'].unique())
print()
print(df_list_orders['city'].unique())
```

```python
# Verify that there are no duplicates in the categorical columns
print(df_order_details['category'].unique())
print()
print(df_order_details['sub_category'].unique())
print()
print(df_sales_target['category'].unique())
```

```python
print(df_order_details.describe())
```

```python
# =====================================
# Date parsing
# =====================================

# Check the type
print(df_list_orders['order_date'].dtype)
print()
print(df_sales_target['month_of_order_date'].dtype)

# Change of column type for date
df_list_orders['order_date'] = pd.to_datetime(df_list_orders['order_date'], dayfirst=True,errors='coerce')

df_sales_target['month_of_order_date'] = pd.to_datetime(df_sales_target['month_of_order_date'], format="%m/%Y",errors='coerce')

print()
print("Fix applied")
print(df_list_orders['order_date'].dtype)
print()
print(df_sales_target['month_of_order_date'].dtype)
```
